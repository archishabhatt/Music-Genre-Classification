# 1. Exploratory Data Analysis

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
df = pd.read_csv("data/features_30_sec.csv")
df.describe()
```

### Correlation Matrix

```python

import seaborn as sns

# Calculate correlation matrix for numeric columns
corr_matrix = df.select_dtypes(include=[np.number]).corr()

# Plot the heatmap
plt.figure(figsize=(20, 15))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Audio Features')
plt.show()
```

The analysis of spectral features revealed strong correlations among the mean values of spectral centroid, spectral bandwidth, and spectral rolloff, with correlation coefficients exceeding 0.9. Additionally, a notable negative correlation of -0.9 was observed between the mean of the second Mel-Frequency Cepstral Coefficient (MFCC2) and the mean values of spectral features.

Scaling down the correlation matrix to 'hero features' to make the plot easier to read and interpret.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

hero_features = [
    'tempo', 'rms_mean', 'spectral_centroid_mean', 
    'spectral_bandwidth_mean', 'rolloff_mean', 
    'zero_crossing_rate_mean', 'chroma_stft_mean', 
    'chroma_stft_var', 'mfcc1_mean', 'mfcc2_mean', 'mfcc3_mean'
]

# 2. Filter and calculate correlation
# Assuming your dataframe is named 'df'
presentation_corr = df[hero_features].corr()

# 3. Create the Matrix
plt.figure(figsize=(12, 10))

# Mask the upper triangle for visual clarity
mask = np.triu(np.ones_like(presentation_corr, dtype=bool))

# Using 'RdBu_r' (Red-Blue Reversed) 
# Red = High Correlation (1.0)
# Blue = High Negative Correlation (-1.0)
# White = No correlation (0.0)
sns.heatmap(presentation_corr, 
            mask=mask,
            cmap='RdBu_r', 
            center=0,
            vmin=-1, vmax=1,
            annot=True, 
            fmt=".2f",
            linewidths=0,
            cbar_kws={"shrink": .8})

# Spotify Branding: Dark Background
plt.gcf().set_facecolor('#191414') 
plt.gca().set_facecolor('#191414')

plt.title('Correlation Matrix', fontsize=22, pad=25, color='white')
plt.xticks(rotation=45, ha='right', color='white', fontsize=12)
plt.yticks(color='white', fontsize=12)

plt.show()
```

### Dendrogram

The dendrogram provides visual insights into the hierarchical clustering of features.

```python
from scipy.cluster import hierarchy
from scipy.spatial.distance import pdist

# Use the correlation matrix to find distances between features
dist = pdist(corr_matrix)
linkage = hierarchy.linkage(dist, method='complete')

plt.figure(figsize=(20, 10))
hierarchy.dendrogram(linkage, labels=corr_matrix.columns, leaf_rotation=90)
plt.title('Hierarchical Clustering Dendrogram of Features')
plt.show()
```

# 2. Principal Component Analysis

Principal Component Analysis (PCA) is a dimensionality reduction technique that transforms a large set of correlated variables into a smaller set of uncorrelated variables called principal components, which retain most of the original data's variance. For this music genre classification project, it was used to simplify the high-dimensional GTZAN feature set, making it easier to visualize clusters of different genres and reducing noise that could lead to overfitting.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler

# Will have to scale the data before applying PCA, due to the different ranges of features.
X = df.drop(['label', 'filename', 'length'], axis=1)
X_scaled = StandardScaler().fit_transform(X)

# Perform PCA
U, s, Vt = np.linalg.svd(X_scaled)
V = Vt.T

# Create figure with both scree and biplot
fig = plt.figure(figsize=(15, 6))

# 1. Scree plot
plt.subplot(121)
var_exp = s**2 / np.sum(s**2)
cum_var_exp = np.cumsum(var_exp)

plt.plot(range(1, len(var_exp) + 1), var_exp, 'bo-', label='Individual')
plt.plot(range(1, len(cum_var_exp) + 1), cum_var_exp, 'ro-', label='Cumulative')
plt.xlabel('Principal Component')
plt.ylabel('Proportion of Variance Explained')
plt.title('Scree Plot')
plt.legend()
plt.grid(True)

# 2. Biplot
plt.subplot(122)
scores = X_scaled @ V
scale = 3  # Adjust this to change relative scaling of arrows

# Plot scores
plt.scatter(scores[:,0], scores[:,1], c='b', alpha=0.5, label='Music Samples')

# Plot loadings
for i, feature in enumerate(X.columns):
    x = V[i,0] * s[0] * scale
    y = V[i,1] * s[1] * scale
    
    plt.arrow(0, 0, x, y, color='r', alpha=0.5, head_width=0.1)
    
    # Add labels with offset based on quadrant
    if x >= 0:
        ha = 'left'
    else:
        ha = 'right'
    if y >= 0:
        va = 'bottom'
    else:
        va = 'top'
        
    plt.text(x*1.1, y*1.1, feature, ha=ha, va=va)

plt.xlabel(f"PC1 ({var_exp[0]:.1%} variance)")
plt.ylabel(f"PC2 ({var_exp[1]:.1%} variance)")
plt.title('PCA Biplot')
plt.grid(True)

# Add legend
plt.plot([0], [0], 'r-', label='Features')
plt.legend()

plt.tight_layout()

# Print feature loadings for reference
print("\nFeature loadings (scaled by singular values):")
for name, v1, v2, v3 in zip(X.columns, 
                       V[:,0] * s[0], 
                       V[:,1] * s[1], 
                       V[:,2] * s[2]):
    print(f"{name:25s}: PC1 = {v1:7.4f}, PC2 = {v2:7.4f}, PC3 = {v3:7.4f}")
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Preprocessing: Scale the data (excluding non-numeric columns like filename)
features = df.drop(columns=['filename', 'label']) if 'label' in df.columns else df
scaler = StandardScaler()
scaled_df = scaler.fit_transform(features.select_dtypes(include=[np.number]))

# Apply PCA
pca = PCA(n_components=2)
pca_res = pca.fit_transform(scaled_df)

# Create a clean plot
plt.figure(figsize=(12, 8))
# Assuming 'label' column exists for genres
if 'label' in df.columns:
    sns.scatterplot(x=pca_res[:, 0], y=pca_res[:, 1], hue=df['label'], palette='viridis', alpha=0.7)
    plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)
else:
    plt.scatter(pca_res[:, 0], pca_res[:, 1], alpha=0.7)

plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
plt.title('PCA of Music Genres')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import numpy as np
import pandas as pd

# Build a dataframe of loadings
loadings_df = pd.DataFrame({
    'feature': X.columns,
    'PC1': V[:,0] * s[0],
    'PC2': V[:,1] * s[1],
    'PC3': V[:,2] * s[2]
})

# Print top 5 per PC sorted by absolute value descending
for pc in ['PC1', 'PC2', 'PC3']:
    print(f"\nTop 5 features for {pc} (sorted by absolute loading):")
    sorted_pc = loadings_df[['feature', pc]].reindex(
        loadings_df[pc].abs().sort_values(ascending=False).index
    ).head(5)
    print(sorted_pc.to_string(index=False))
```

Simplified classification structure by reassigning the 10 original GTZAN genres into three broader categories — **Accoustic**, **Rhythmic** and **Electric**, by focusing on similar spectral characteristics and instrumentation.

```python
acoustic = ['classical', 'blues', 'jazz']
rhythmic = ['country', 'reggae', 'rock', 'hiphop']
electric = ['disco', 'pop', 'metal']

def assign_group(label):
    if label in acoustic:
        return 'Acoustic'
    elif label in rhythmic:
        return 'Rhythmic'
    elif label in electric:
        return 'Electric'

df['final_group'] = df['label'].apply(assign_group)
```
